```python
import numpy as np
from sklearn.feature_selection import VarianceThreshold
```

```python
# Carga de datos
X = [[0, 0, 1], [0, 1, 0], [1, 0, 0],
     [0, 1, 1], [0, 1, 0], [0, 1, 1]]

print(X)
```

```python
# Limpieza de datos: seleccion de atributos

# Fijamos un umbral (th) para decidir que porcentaje minimo deben variar los valores de una columna para considerarse
# una columna relevate a la hora de mantenerla en nuestro estudio

th = 0.2 #Limite de decision
sel = VarianceThreshold(th)
print(sel.fit_transform(X))
```
